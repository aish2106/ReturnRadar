```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df['item_price'], bins=30, edgecolor='black')
plt.title('Distribution of Item Prices')
plt.xlabel('Item Price')
plt.ylabel('Number of Products')
plt.show()
```

##### The histogram shows that item prices are right-skewed. Most products are concentrated within the lower-to-middle price range, approximately 20–60. A smaller number of products have higher prices, extending up to 259.90. This indicates the presence of potentially extreme values that require further investigation.

```python
plt.figure(figsize=(8, 4))
plt.boxplot(df['item_price'])
plt.title('Boxplot of Item Prices')
plt.ylabel('Item Price')
plt.show()
```

```python
Q1 = df['item_price'].quantile(0.25)
Q3 = df['item_price'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[df['item_price'] > upper_bound]

print("Number of outliers:", len(outliers))
print("Outlier percentage:", len(outliers) / len(df) * 100)
```

```python
print("Outlier return distribution:")
print(outliers['return'].value_counts())

print("\nReturn rate among outliers:")
print(outliers['return'].mean() * 100)
```

```python
normal_products = df[df['item_price'] <= upper_bound]

print("Return rate of normal products:")
print(normal_products['return'].mean() * 100)

print("\nReturn rate of outlier products:")
print(outliers['return'].mean() * 100)
```

#### Categorical column:

```python
print(df['item_size'].value_counts())
```

```python
print(sorted(df['item_size'].dropna().unique()))
```

```python
size_return = df.groupby('item_size')['return'].agg(
    ['count', 'mean'])

size_return['return_rate'] = size_return['mean'] * 100

size_return = size_return.sort_values(
    'return_rate', ascending=False)

print(size_return)
```

```python
size_return_reliable = size_return[
    size_return['count'] >= 50]

print(size_return_reliable.sort_values(
    'return_rate', ascending=False))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.bar(
    size_return_reliable.index,
    size_return_reliable['return_rate']
)

plt.title('Return Rate by Item Size')
plt.xlabel('Item Size')
plt.ylabel('Return Rate (%)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

##### Item size analysis revealed differences in return rates across size categories. Some numerical sizes showed return rates above 55%, while commonly occurring alphabetical sizes showed rates around 41–44%. Unsized products had a lower return rate. These patterns suggest that item size may be a useful feature to investigate during return-risk modelling.
